```
// 5. Get tenants whose Name contains a given substring, case-insensitively.

string substr = "cOn";

var tenantsNameContainsSubstring 
    = await db.Tenants.Where(a => a.Name.ToUpper().Contains(substr.ToUpper())).ToListAsync();
tenantsNameContainsSubstring
```

```
// 7. Get the count of all permissions grouped by Module.

var countAllPermissions 
    = await db.Permissions
        .GroupBy(e => e.Module)
        .Select(g => new {
            Module = g.Key,
            Count = g.Count()
        }).ToListAsync();

countAllPermissions
```

```
// 8. Get distinct Module values from Permission.

var distncitModules 
    = await db.Permissions
        .Select(a => a.Module)
        .Distinct()
        .ToListAsync();

distncitModules
```

```
// 9. Check whether a username already exists (AnyAsync) before insert.

var isUserExists
    = await db.Users
        .AnyAsync(a => a.Username == "ujjwal");

isUserExists
```

```
// 10. Get users ordered by CreatedAt, then break ties with UserId.

var users = await db.Users
                .OrderByDescending(a => a.CreatedAt)
                .ThenByDescending(a => a.UserId)
                .ToListAsync();
users
```

```
// 11. Get users whose email domain is contoso.io (parse from Email).

var users = await db.Users
                .Where(a => a.Email.Substring(a.Email.IndexOf("@")+1) == "contoso.io")
                .ToListAsync();
users
```

```
// 12. Get users where AvatarUrl is null OR RefferedById is null.

var users = await db.Users
                    .Where(a => a.AvatarUrl == null || a.RefferedById == null)
                    .ToListAsync();
users
```

```
// 13. Get tenants expiring in the next 30 days AND still active.

var tenants = await db.Tenants
                        .Where(a => a.ExpireAt < DateTime.Today.AddDays(30) && a.ExpireAt > DateTime.Today && a.IsActive)
                        .ToListAsync();
tenants
```

## Joins (21–35)

```
// 21. Join Users to Tenants on TenantId (no FK exists — write it as an explicit join).

var result = await db.Users
                .Join(
                    db.Tenants,
                    user => user.TenantId,
                    tenant => tenant.Id,
                    (user, tenant) => new
                        {
                            UserId = user.UserId,
                            Username = user.Username,
                            TenantId = tenant.Id,
                            TenantName = tenant.Name
                        }
                )
                .ToListAsync();
result
```

```
// 22. Join UserRole → Role → RolePermission → Permission to list a user's effective permissions.

// select permissionId, permissionName from UserRole ur
// join Role r on r.Id = ur.RoleId
// join RolePermission rp on rp.RoleId = r.Id
// join Permission p on p.Id = rp.PermissionId

var permissions = (
    from ur in db.UserRoles
    join r in db.Roles
        on ur.RoleId equals r.Id 
    join rp in db.RolePermissions
        on r.Id equals rp.RoleId 
    join p in db.Permissions
        on rp.PermissionId equals p.Id 
    select new
    {
        PermissionId = p.Id,
        PermissionName = p.Name,
    }
).ToList();

permissions
```
